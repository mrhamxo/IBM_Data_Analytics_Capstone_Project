# IBM Data Analyst Capstone Project
## Module 4 — Exploratory Data Analysis (EDA)

**Goal:** Explore the cleaned survey dataset to understand distributions,  
detect outliers, and reveal relationships between variables such as  
salary, education level, age, and work location.

```python
# ── CELL 1: Import libraries ───────────────────────────────────────────────────
import pandas as pd      # data manipulation
import numpy as np       # numerical calculations
import matplotlib.pyplot as plt   # base plotting
import seaborn as sns    # statistical visualisation
import os

# Apply a clean visual style for all plots
sns.set_theme(style="whitegrid")
plt.rcParams["figure.dpi"] = 120
```

```python
# ── CELL 2: Load the cleaned dataset (from Module 3) ──────────────────────────
# Fall back to the IBM-hosted URL if the local file doesn't exist yet
local_path = os.path.join("..", "data", "survey_data_cleaned.csv")
fallback_url = (
    "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/"
    "IBM-DA0321EN-SkillsNetwork/LargeData/m2_survey_data.csv"
)

if os.path.exists(local_path):
    df = pd.read_csv(local_path)
    print("Loaded local cleaned file.")
else:
    df = pd.read_csv(fallback_url)
    print("Loaded fallback URL.")

print("Shape:", df.shape)
df.head()
```

```python
# ── CELL 3: Basic descriptive statistics ─────────────────────────────────────
# .describe() shows count, mean, std, min, quartiles, and max for numeric columns
df.describe()
```

```python
# ── CELL 4: Distribution of annual compensation (salary) ─────────────────────
# Remove extreme outliers for a cleaner plot (keep values below 99th percentile)
salary_99 = df["ConvertedComp"].quantile(0.99)
df_salary  = df[df["ConvertedComp"] <= salary_99]

print(f"99th percentile salary cutoff: ${salary_99:,.0f}")
print(f"Median salary: ${df['ConvertedComp'].median():,.0f}")
print(f"Mean salary:   ${df['ConvertedComp'].mean():,.0f}")
```

```python
# ── CELL 5: Age distribution ──────────────────────────────────────────────────
# Understand the age profile of survey respondents
print("Age statistics:")
print(df["Age"].describe())
print(f"\nMost common age group: {df['Age'].mode()[0]}")
```

```python
# ── CELL 6: Count of respondents by education level ──────────────────────────
# Value counts give us absolute frequency per category
ed_counts = df["EdLevel"].value_counts()
print("Respondents by education level:")
print(ed_counts.to_string())
```

```python
# ── CELL 7: Work location distribution ────────────────────────────────────────
workloc_counts = df["WorkLoc"].value_counts()
print("Work location distribution:")
print(workloc_counts.to_string())
```

```python
# ── CELL 8: Correlation between age and compensation ─────────────────────────
# Pearson correlation coefficient — closer to 1 or -1 means stronger relationship
corr = df[["Age", "ConvertedComp"]].corr()
print("Correlation matrix (Age vs Salary):")
print(corr)
```

```python
# ── CELL 9: Median salary by education level ──────────────────────────────────
# Group by education, compute median salary, and sort descending
salary_by_ed = (
    df.groupby("EdLevel")["ConvertedComp"]
    .median()
    .sort_values(ascending=False)
)
print("Median annual salary (USD) by education level:")
print(salary_by_ed.to_string())
```

```python
# ── CELL 10: Outlier check — IQR method for ConvertedComp ────────────────────
Q1  = df["ConvertedComp"].quantile(0.25)   # first quartile
Q3  = df["ConvertedComp"].quantile(0.75)   # third quartile
IQR = Q3 - Q1                              # interquartile range

lower_fence = Q1 - 1.5 * IQR
upper_fence = Q3 + 1.5 * IQR

outliers = df[(df["ConvertedComp"] < lower_fence) | (df["ConvertedComp"] > upper_fence)]
print(f"IQR: ${IQR:,.0f}")
print(f"Lower fence: ${lower_fence:,.0f}")
print(f"Upper fence: ${upper_fence:,.0f}")
print(f"Outlier rows: {len(outliers)} ({len(outliers)/len(df)*100:.1f}%)")
```
